```python
def dms_to_decimaldeg (degrees, minutes, seconds):
    if degrees < 0:
        result = degrees - minutes/60 - seconds/3600
    else:
        result = degrees + minutes/60 + seconds/3600
    return result

```

```python
# Exercise: 
''' 
Given a dict with three destination cities and their coordinates. I have to develop a for loop
to iterate over the destination_cities dict and call a get_driving_distance() function to print
real driving distance between San Francisco and each city.

I need also to rate limit my requests by additing time.sleep(2) between successful function calls
'''

# I have already imported requests and time libraries/modules

def get_driving_distance(source_coordinates, dest_coordinates):
    parameters = {
        'api_key' : ORS_API_KEY,
        'start' : f'{source_coordinates[1]}, {source_coordinates[0]}',
        'end' : f'{dest_coordinates[1]}, {dest_coordinates[0]}'
    }
    response = requests.get(
        'https://api.heigit.org/openrouteservice/v2/directions/driving-car?',params=parameters
    )
    if response.status_code == 200:
        data = response.json()
        summary = data['features'][0]['properties']['summary']
        distance = summary['distance']
        return distance/1000
    else:
        print('Request failed')
        print('Reason: ', response.text)
        return -9999

san_francisco = (37.7749, -122.4194)

destination_cities = {
    'Los Angeles' : (34.0522, -118.2437),
    'Boston' : (42.3601, -71.0589),
    'Atlanta' : (33.7490, -84.3880)
}

# Calculating distance for the three cities:

for city in destination_cities:
    dest_coordinates = destination_cities[city]
    source_coordinates = san_francisco
    distance = get_driving_distance(source_coordinates, dest_coordinates)
    print(f'Driving distance between {city} and San Francisco is: ', distance)
    time.sleep(2)
```

```python
# Geocoding addresses using GeoPy

''' 
Given a tuple of place names along with their address. I need to obtain their lat and long
I will need to use Nominatim geocoder and obtain the lat and log of each address
'''

# List of hurricane evac centres in NYC
# Each item is a tuple with the name of the centre and its address

locations = [
    ('Norman Thomas HS (ECF)', '111 E 33rd St, NYC, New York'),
    ('Midtown East Campus', '233 E 56th St, NYC, New York'),
    ('Louis D. Brandeis HS', '145 W 84th St, NYC, New York'),
    ('Martin Luther King, Jr. HS', '122 Amsterdam Avenue, NYC, New York'),
    ('P.S 48', '4360 Broadway, NYC, New York')
]

# Importing Nominatim module from the geopy geocoders library
from geopy.geocoders import Nominatim
 
# Initializing the geolocator
geolocator = Nominatim(user_agent = 'mayerekenford@gmail.com')

# Geocoding the locations from the given list
geocoded_list = []
for centre in locations:
    location = geolocator.geocode(centre[1])
    match = tuple({
        'centre': centre[0],
        'lat': location.latitude,
        'lon': location.longitude}.values())
    geocoded_list.append(match)

geocoded_list
```

```python
# Doing a reverse to get the addresses in full
address = []
for addr in geocoded_list:
    location = geolocator.reverse(f'{addr[1]},{addr[2]}')
    match_addr = tuple({'name': addr[0],
                        'addr': location.address}.values())
    address.append(match_addr)

address
```
